```sql
-- fact_ohlc tabel aanmaken (open, high, low, close) --
CREATE TABLE IF NOT EXISTS gold.fact_ohlc(
      coin_id STRING,
      open_price DOUBLE,
      high_price DOUBLE,
      low_price DOUBLE,
      close_price DOUBLE,
      date DATE
);

-- fact_ohlc vullen met unieke waarden uit de silver tabel --
INSERT INTO gold.fact_ohlc
WITH daily AS ( -- CTE voor dagelijkse O-H-L-C waarden uit silver tabel
  SELECT 
    coin_id,
    FIRST(price) AS open_price,
    MAX(price) AS high_price,
    MIN(price) AS low_price,
    LAST(price) AS close_price,
    date
  FROM silver.crypto_prices
  GROUP BY coin_id, date
)
SELECT d.* 
FROM daily d
LEFT ANTI JOIN gold.fact_ohlc g
ON d.coin_id = g.coin_id 
AND d.date = g.date;

-- resultaat
SELECT * FROM gold.fact_ohlc;
```
