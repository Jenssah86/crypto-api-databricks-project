```sql
-- Stored Procedure voor dim_coin
CREATE OR REPLACE PROCEDURE gold.load_dim_coin()
SQL SECURITY INVOKER 
BEGIN
  INSERT INTO gold.dim_coin
  SELECT DISTINCT s.coin_id
  FROM silver.crypto_prices s
  LEFT ANTI JOIN gold.dim_coin d -- alleen unieke waarden toevoegen
  ON s.coin_id = d.coin_id;
END;

-- Stored Procedure voor dim_date
CREATE OR REPLACE PROCEDURE gold.load_dim_date()
SQL SECURITY INVOKER
BEGIN
  INSERT INTO gold.dim_date
  SELECT DISTINCT s.date,
   EXTRACT(WEEK FROM date) as week,
   EXTRACT(MONTH FROM date) as month,
   EXTRACT (QUARTER FROM date) as quarter,
   EXTRACT(YEAR FROM date) as year
  FROM silver.crypto_prices s
  LEFT ANTI JOIN gold.dim_date d -- alleen unieke waarden toevoegen
  ON s.date = d.date;
END;
```
