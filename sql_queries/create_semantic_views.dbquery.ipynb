```sql
-- View voor dim_coin

CREATE OR REPLACE VIEW semantic.dim_coin AS
SELECT
  coin_id
FROM gold.dim_coin;

-- View voor dim_date

CREATE OR REPLACE VIEW semantic.dim_date AS
SELECT
  date,
  week,
  month,
  quarter,
  year
FROM gold.dim_date;

-- View voor fact_ohlc (hoofdfact)

CREATE OR REPLACE VIEW semantic.fact_ohlc AS
SELECT
  coin_id,
  date,
  open_price,
  high_price,
  low_price,
  close_price
FROM gold.fact_ohlc;

-- View voor fact_prices (alleen intraday)

CREATE OR REPLACE VIEW semantic.fact_prices AS
SELECT
  coin_id,
  price_timestamp,
  price
FROM gold.fact_prices;
```
