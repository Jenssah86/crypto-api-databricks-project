```sql
-- dim_coin tabel aanmaken --
CREATE TABLE IF NOT EXISTS gold.dim_coin (
    coin_id STRING
);

-- dim_coin vullen met unieke waarden uit de silver tabel --
INSERT INTO gold.dim_coin
SELECT DISTINCT s.coin_id FROM silver.crypto_prices s
LEFT ANTI JOIN gold.dim_coin g ON s.coin_id = g.coin_id
WHERE s.coin_id IS NOT NULL;

-- resultaat
SELECT * FROM gold.dim_coin;
```
