```sql
-- fact_prices tabel aanmaken --
CREATE TABLE IF NOT EXISTS gold.fact_prices(
      coin_id STRING,
      price DOUBLE,
      price_timestamp TIMESTAMP,
      timestamp_unix BIGINT, -- timestamp in unix voor performance/eenduidigheid
      date DATE
);

-- fact_prices vullen met unieke waarden uit de silver tabel --
INSERT INTO gold.fact_prices
SELECT DISTINCT
      s.coin_id,
      s.price,
      s.price_timestamp,
      unix_timestamp(s.price_timestamp) AS timestamp_unix,
      s.date
FROM silver.crypto_prices s
LEFT ANTI JOIN gold.fact_prices g
ON s.coin_id = g.coin_id 
AND s.price_timestamp = g.price_timestamp;

-- resultaat
SELECT * FROM gold.fact_prices;
```
